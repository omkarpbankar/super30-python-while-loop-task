# While Loop Task

This notebook practices condition-controlled iteration. A `while` loop is useful when the number of repetitions depends on a condition, a sentinel value, or user input rather than a known sequence.

Each major loop labels its **initialization**, **condition**, and **update/termination** so that the loop has a clear stopping path.

## 1. Print numbers from 1 to 100

A `while` loop shows the three control parts explicitly: start at 1, continue through 100, and increase by 1.

```python
number = 1  # Initialization
while number <= 100:  # Condition
    print(number, end=" " )
    number += 1  # Update
```

## 2. Print numbers from 100 to 1

```python
number = 100  # Initialization
while number >= 1:  # Condition
    print(number, end=" " )
    number -= 1  # Update
```

## 4. Sum the digits of an integer

The last digit is found with `% 10`; removing it is done with `// 10`. The loop ends when no digits remain.

```python
def sum_of_digits(number):
    number = abs(number)  # Initialization
    digit_sum = 0
    while number > 0:  # Condition
        digit_sum += number % 10
        number //= 10  # Update: remove the last digit
    return digit_sum

print(sum_of_digits(5832))  # 18
```

## 5. Reverse an integer using `while`

```python
def reverse_integer(number):
    sign = -1 if number < 0 else 1
    number = abs(number)  # Initialization
    reversed_number = 0
    while number > 0:  # Condition
        reversed_number = reversed_number * 10 + number % 10
        number //= 10  # Update: remove the last digit
    return sign * reversed_number

print(reverse_integer(12345))  # 54321
```

## 6. Count the digits in an integer

```python
def count_digits(number):
    number = abs(number)  # Initialization
    if number == 0:
        return 1
    count = 0
    while number > 0:  # Condition
        count += 1
        number //= 10  # Update: remove the last digit
    return count

print(count_digits(5832))  # 4
```

## 7. Calculate factorial using `while`

```python
def factorial(number):
    if number < 0:
        raise ValueError("Factorial is undefined for negative numbers.")
    result = 1  # Initialization
    current = 1
    while current <= number:  # Condition
        result *= current
        current += 1  # Update
    return result

print(factorial(5))  # 120
```

## 8. Sum numbers until the user enters 0

Here `0` is a sentinel value. This is a natural `while` problem because the number of entries is unknown.

```python
def sum_until_zero():
    total = 0  # Initialization
    number = None
    while number != 0:  # Condition
        number = int(input("Enter a number (0 to stop): "))
        if number != 0:
            total += number
        # Termination occurs when number becomes 0.
    print("Sum:", total)

# sum_until_zero()  # Uncomment to run interactively
```

## 9. Password checker

```python
def password_checker(correct_password):
    password = ""  # Initialization
    while password != correct_password:  # Condition
        password = input("Enter password: " )
        if password != correct_password:
            print("Incorrect password. Try again.")
        # Termination occurs when password matches.
    print("Access granted.")

# password_checker("python123")  # Uncomment to run interactively
```

## 10. Guessing game

```python
def guessing_game(secret_number):
    guess = None  # Initialization
    while guess != secret_number:  # Condition
        guess = int(input("Guess the number: "))
        if guess < secret_number:
            print("Too low.")
        elif guess > secret_number:
            print("Too high.")
        # Termination occurs when guess equals secret_number.
    print("Correct!")

# guessing_game(37)  # Uncomment to run interactively
```

## 11. Menu-driven calculator

The menu repeats until option 5 is selected. Division by zero is handled explicitly.

```python
def calculator():
    choice = 0  # Initialization
    while choice != 5:  # Condition
        print("\n1. Add\n2. Subtract\n3. Multiply\n4. Divide\n5. Exit")
        choice = int(input("Choose an option: "))
        if choice == 5:
            print("Calculator closed.")
            break  # Termination
        if choice not in (1, 2, 3, 4):
            print("Invalid option.")
            continue  # Update/termination path: return to the menu
        first = float(input("First number: "))
        second = float(input("Second number: "))
        if choice == 1:
            print("Result:", first + second)
        elif choice == 2:
            print("Result:", first - second)
        elif choice == 3:
            print("Result:", first * second)
        elif second == 0:
            print("Cannot divide by zero.")
        else:
            print("Result:", first / second)
        # The next menu selection updates choice; selecting 5 terminates.

# calculator()  # Uncomment to run interactively
```

```python
def calculator_demo():
    scripted_choices = [1, 2, 5]
    index = 0  # Initialization
    while index < len(scripted_choices):  # Condition
        choice = scripted_choices[index]
        if choice == 1:
            print("12 + 8 =", 12 + 8)
        elif choice == 2:
            print("12 - 8 =", 12 - 8)
        elif choice == 5:
            print("Exit selected.")
            break  # Termination
        index += 1  # Update

calculator_demo()  # Demonstrates Add, Subtract, and Exit
```

## 12. ATM menu using `while`

The balance is state that changes across menu selections. The application continues until Exit is chosen.

```python
def atm_menu():
    balance = 1000.0  # Initialization
    choice = 0
    while choice != 4:  # Condition
        print("\n1. Check balance\n2. Deposit\n3. Withdraw\n4. Exit")
        choice = int(input("Choose an option: "))
        if choice == 1:
            print("Balance:", balance)
        elif choice == 2:
            balance += float(input("Deposit amount: "))
            print("Updated balance:", balance)
        elif choice == 3:
            amount = float(input("Withdrawal amount: "))
            if 0 < amount <= balance:
                balance -= amount
                print("Updated balance:", balance)
            else:
                print("Invalid amount or insufficient funds.")
        elif choice == 4:
            print("Thank you.")
        else:
            print("Invalid option.")
        # choice is updated by the next input; 4 is the termination option.

# atm_menu()  # Uncomment to run interactively
```

## Why `while` is preferable here

A `for` loop is a natural choice when iterating through a known sequence or fixed range. These programs use `while` because the stopping point is controlled by a changing condition: a sentinel `0`, a matching password, a correct guess, or a menu choice of Exit. The loop can continue for any number of attempts without guessing the count in advance.
